# Assignment 3 – Encoding Categorical Variables

## Objective
To convert categorical variables into numerical form using LabelEncoder and OneHotEncoder.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
data = pd.read_csv("../Assignment_1_Data_Analysis/Titanic-Dataset.csv")
```

```python
data.head()
```

```python
data['Sex'].unique()
```

```python
le = LabelEncoder()

data['Sex'] = le.fit_transform(data['Sex'])
```

```python
data[['Sex']].head()
```

```python
dict(zip(le.classes_, le.transform(le.classes_)))
```

```python
data['Embarked'].unique()
```

```python
data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0])
```

```python
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
```

```python
embarked_encoded = ohe.fit_transform(data[['Embarked']])
```

```python
embarked_df = pd.DataFrame(
    embarked_encoded,
    columns=ohe.get_feature_names_out(['Embarked'])
)
```

```python
embarked_df.head()
```

```python
data = data.drop('Embarked', axis=1)
```

```python
data = pd.concat([data, embarked_df], axis=1)
```

```python
data.head()
```

```python
data.dtypes
```

## Conclusion

Categorical variables in the Titanic dataset were converted into numerical form. LabelEncoder was used for the binary Sex variable, while OneHotEncoder was used for the categorical Embarked variable. The encoded dataset can now be used by machine learning algorithms that require numerical input.
