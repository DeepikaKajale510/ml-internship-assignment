# Assignment 2 – Handling Missing Data

## Objective
To identify missing values in the Titanic dataset and handle them using mean/median/mode imputation.

```python
import pandas as pd
```

```python
data = pd.read_csv("../Assignment_1_Data_Analysis/Titanic-Dataset.csv")
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
median_age = data['Age'].median()

data['Age'] = data['Age'].fillna(median_age)
```

```python
data['Age'].isnull().sum()
```

```python
data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0])
```

```python
data['Embarked'].isnull().sum()
```

```python
data.isnull().sum()
```

## Conclusion

Missing values in the Titanic dataset were identified and handled using appropriate imputation techniques. Median imputation was used for the numerical Age column, while mode imputation was used for the categorical Embarked column.
